BINARY

```python
def binarysearc(n,arr,item):
    start = 0
    end = n-1
    mid = (start+end)//2
    while(start <= end):
        if (arr[mid] == item):
            return True
        elif(item > arr[mid]):
            start = mid + 1
            mid = (start+end)//2
        else:
            end = mid - 1
            mid = (start+end)//2
    return False
n = int(input("ENTER THE LENGTH OF ARRAY : "))
arr = list(map(int,input("ENTER THE ARRAY : ").split()))
item = int(input("ENTER THE ITEM : "))

x = binarysearc(n,arr,item)
print(x)
```

REVERSE A STRING

```python
def rever(s):
    s1 = ""
    i = len(s)-1
    while(i>=0):
        s1 = s1 + s[i]
        i = i - 1
    return s1
s = input()
print(rever(s))
```

```python
def reverseee(s):
    s = list(s)
    start = 0
    end = len(s)-1
    while(start < end):
        s[start],s[end] = s[end],s[start]
        start = start + 1
        end = end - 1
    s = "".join(s)
    return s
s = input()
print(reverseee(s))
```

BUBBLE SORT

```python
def bubble(n,arr):
    for i in range(0,n):
        swapped = 0
        for j in range(0,n-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swapped = 1
        if swapped == 0:
            break 
    return arr
n = int(input("ENTER N : "))
arr = list(map(int,input().split()))
print(bubble(n,arr))
```

SUM OF PAIRS == K

```python
def pairsum(n,arr,k):
    count = 0
    for i in range(0,n):
        for j in range(i+1,n):
            if arr[i]+arr[j] == k:
                count = count + 1
    return count
n = int(input())
arr = list(map(int,input().split()))
k = int(input())
print(pairsum(n,arr,k))
```

```python
from math import sqrt
def prime(n):
    if n < 2 :
        return False
    i = 2
    end = int(sqrt(n))
    while(i <= end):
        if n%i == 0:
            return False
        i = i + 1
    return True
def pairsum2(n,arr):
    count = 0
    for i in range(0,n):
        for j in range(i+1,n):
            if prime(arr[i]) == True and prime(arr[j]) == True:
                count = count + 1
    return count
n = int(input())
arr = list(map(int,input().split()))
print(pairsum2(n,arr))
```

SECOND HIGHEST ELEMENT

```python
def secondHighest(n,arr):
    highest = float('-inf')
    second = float('-inf')

    for i in range(0,n):
        if arr[i] > highest:
            second = highest
            highest = arr[i]
        elif arr[i] > second and arr[i] != highest:
            second = arr[i]
    return second
n = int(input())
arr = list(map(int,input().split()))
print(secondHighest(n,arr))
```

```python
def fun(n,arr):
    if n == 1:
        return n
    inx = 1
    for i in range(1,n):
        if arr[i] != arr[i-1]:
            arr[inx] = arr[i]
            inx = inx + 1
    return inx
n = int(input())
arr = list(map(int,input().split()))
x = fun(n,arr)
for i in range(0,x):
    print(arr[i],end = " ")
        
```
