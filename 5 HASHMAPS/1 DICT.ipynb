REMOVE DUPLICATES FROM SORTED AND UNSORTED ARRAY

```python
def duplicates(arr):
    d = {}
    for i in arr:
        d[i] = 0
    for k in d.keys():
        print(k)
arr = list(map(int,input().split()))
duplicates(arr)
```

REMOVE DUPLICATES FROM STRING

```python
def remove_duplicates(s):
    d = {}
    for i in s:
        d[i] = 0
    s1 = "" 
    for k in d.keys():
        s1 = s1 + k
    return s1
s = input()
print(remove_duplicates(s))
```

```python
def remove_duplicates_2(s):
    d = {}
    for i in s:
        d[i] = 0
    for k in d.keys():
        print(k,end="")
s = input()
remove_duplicates_2(s)
```

COUTING FREQUENCY OF EACH CHARACTER IN STRING

```python
def freq_count(s):
    d = {}
    for i in s:
        d[i] = 0
    for i in s:
        d[i] = d[i] + 1
    for k,v in d.items():
        print(k," occured ",v," times")
s = input()
freq_count(s)
```

FIND DUPLICATES

```python
def find_duplicates(s):
    d = {}
    for i in s:
        d[i] = 0
    for i in s:
        d[i] = d[i] + 1
    for k,v in d.items():
        if v > 1:
            print(k,end="")
s = input()
find_duplicates(s)
```

SUM OF PAIRS == TARGET

```python
def isexists(arr,target):
    s = set()
    for num in arr:
        component = target - num
        if component in s:
            return True
        s.add(num)
    return False
n = int(input())
arr = list(map(int,input().split()))
target = int(input())
x = isexists(arr,target)
print(x)
```

PRINT OCCURANCE OF EACH ELEMENT IN DECENDING ORDER

```python
arr = list(map(int,input().split()))
d = {}
for i in arr:
    d[i] = 0
for i in arr:
    d[i] = d[i] + 1
d = sorted(d.items(),key = lambda x : x[1]) # x[1] means values and x[0] means keys
d = dict(d)                                 # so we want to sort according to keys then write x[0]
for k,v in d.items():
    print(k , "OCCURED", v)
```

ASCENDING ORDER

```python
arr = list(map(int,input().split()))
d = {}
for i in arr:
    d[i] = 0
for i in arr:
    d[i] = d[i] + 1
d = sorted(d.items(),key = lambda x : x[1],reverse=True)
d = dict(d)
for k,v in d.items():
    print(k , "OCCURED", v)
```
