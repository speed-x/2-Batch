SWAP 2 VARIABLES WITH 3rd VARIABLE

```python
def swap(a,b):
    c = a
    a = b
    b = c
    print(a,b)
a = int(input("ENTER FIRST NUMBER : "))
b = int(input("ENTER SECOND NUMBER : "))
swap(a,b)
```

SWAP 2 VARIABLES WITHOUT USING 3rd VAIABLE

```python
def swap(a,b):
    a = a+b
    b = a-b
    a = a-b
    print(a,b)
a = int(input("ENTER FIRST NUMBER : "))
b = int(input("ENTER ANOTHER NUMBER : "))
swap(a,b)
```

SWAP 2 VARIBLES USING PYTHON

```python
def swap(a,b):
    a,b = b,a
    print(a,b)
a = int(input("ENTER A NUMBER : "))
b = int(input("ENTER ANOTHER NUMBER : "))
swap(a,b)
```

SWAP 3 VARIABLES 

```python
def swap(a,b,c):
    a = a+b+c
    b = a-b-c
    c = a-b-c
    a = a-b-c
    print(a,b,c) 
a = int(input("ENTER FIRST NUMBER : "))
b = int(input("ENTER SECOND NUMBER : "))
c = int(input("ENTER THIRD NUMBER : "))
swap(a,b,c)
```

REVERSE A NUMBER

```python
def reverse(n):
    rev = 0
    while(n):
        b = n%10
        rev = (rev*10) + b
        n = n//10
    print(rev)
n = int(input("ENTER A NUMBER : "))
reverse(n)
```

PALINDROM

```python
def palindrom(n):
    rev = 0
    a = n
    while(n):
        b = n%10
        rev = (rev * 10) + b
        n = n//10
    if rev == a:
        print("PALINDROM")
    else:
        print("NOT A PALINDROM")
n = int(input("ENTER A NUMBER : "))
palindrom(n)
```

SUM OF REVERSE AND NUMBER

```python
def rev_sum(n):
    a = n
    rev = 0
    while(n):
        b = n%10
        rev = (rev * 10) + b
        n = n//10
    print(a + rev)
n = int(input("ENTER A NUMBER : "))
rev_sum(n)
```

ARMSTRONG

```python
def countdigits(n):
    count = 0
    while(n):
        n = n//10
        count = count + 1
    return (count)
def arms(n):
    c = countdigits(n)
    a = n
    sum = 0
    while(n):
        b = n%10
        sum = sum + (b**c)
        n = n//10
    if sum == a:
        print("ARMSTRONG")
    else:
        print("NOT A ARMSTRONG")    
n = int(input("ENTER A NUMBER : "))
arms(n)
```

DISARUM

```python
def countdigits(n):
    count = 0
    while(n):
        n = n//10
        count = count + 1
    return (count)
def desa(n):
    c = countdigits(n)
    sum = 0
    a = n
    while(n):
        b = n%10
        sum = sum + (b**c)
        n = n//10
        c = c - 1
    if a == sum:
        print("DIASUM NUMBER")
    else:
        print("NOT A DIASUM NUMBER")
n = int(input("ENTER A NUMBER : "))
desa(n)
```

SPY NUMBER

```python
def spy(n):
    sum = 0
    prod = 1
    while(n):
        b = n%10
        sum = sum + b
        prod = prod * b
        n = n//10
    if sum == prod:
        print("SPY NUMBER")
    else:
        print("NOT A SPY NUMBER")
n = int(input("ENTER A NUMBER"))
spy(n)
```

PERFECT NUMBER

```python
def perfect(n):
    i = 1
    sum = 0
    while(i <= n//2):
        if n%i == 0:
            sum = sum + i
        i = i + 1
    if sum == n:
        print("PERFECT NUMBER")
    else:
        print("NOT A PERFECT NUMBER")
n = int(input("ENTER A NUMBER : "))
perfect(n)
```
