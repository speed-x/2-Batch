SUNNY

```python
from math import sqrt
def sunny(n):
    n = n + 1
    x = sqrt(n)
    y = int(x)
    if x == y:
        return True
    return False
n = 15
sunny(n)
```

HAPPY NUMBER

```python
def happy(n):
    sum = n
    while(sum > 9):
        local = 0
        while(sum != 0):
            b = sum%10
            local = local + (b*b)
            sum = sum//10
        sum = local
    if sum == 1 or sum == 7:
        return True
    return False
n = int(input())
print(happy(n))
```
