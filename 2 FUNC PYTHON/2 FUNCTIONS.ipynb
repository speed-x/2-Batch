PRIME NUMBER

```python
from math import sqrt
def prime(n):
    if n <= 1:
        return False
    i = 2
    end = int(sqrt(n))
    while(i <= end):
        if n%i == 0:
            return False
        i = i + 1
    return True
n = int(input("ENTER A NUMBER : "))
print(prime(n))
```

```python
from math import sqrt
def prime(n):
    if n <= 1:
        return False
    end = int(sqrt(n))
    for i in range(2,end+1):
        if n%i == 0:
            return False
    return True
n = int(input("ENTER A NUMBER : "))
print(prime(n))
```

PRINTING FIRST N TERMS OF FIBONACCI SERIES

```python
def fibo(n):
    if n == 1:
        print(0)
    elif n == 2:
        print(0)
        print(1)
    else:
        print(0)
        print(1)
        n = n - 2
        a = 0
        b = 1
        while(n):
            c = a+b
            print(c)
            a = b
            b = c
            n = n - 1
n = int(input("ENTER A NUMBER : "))
fibo(n)
```

FIND A NUMBER IF IT IS IN A FIBO SERIES

```python
def find(n):
    p = n
    if n == 1 or n == 0:
        return True
    else:
        a = 0
        b = 1
        #n = n - 2
        while(n):
            c = a + b
            if c == p:
                return True
            a = b
            b = c
            n = n - 1
    return False
n = int(input("ENTER A NUMBER : "))
find(n)
```
