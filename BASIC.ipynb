SPY NUMBER

```python
SUM OF DIGITS == PRODUCT OF DIGITS
```

```python
n = int(input("ENTER N : "))
sum = 0
prod = 1
while(n != 0):
    sum = sum + n%10
    prod = prod * (n%10)
    n = n//10
if sum == prod:
    print("SPY NUMBER")
else:
    print("NOT A SPY NUMBER")
```

AMRSTRONG NUMBER

```python
sum of digits with raised to the power of number of digits == number itself
```

```python
n = int(input("ENTER N : "))
c = 0
a = n
while(n != 0):
    c = c + 1
    n = n//10
n = a
num = 0
while(n != 0):
    b = n%10
    num = num + (b ** c)
    n = n//10
if num == a:
    print("ARMS")
else:
    print("NOT ARMS")
```

DISARIUM NUMBER

```python
A number is called Disarium if sum of its digits powered with their respective positions is equal to the number itself.
```

```python
n = int(input("ENTER N : "))
c = 0
a = n
while(n != 0):
    c = c + 1
    n = n//10
n = a
num = 0
while(n != 0):
    b = n%10
    num = num + (b ** c)
    n = n//10
    c = c - 1
if num == a:
    print("DIA")
else:
    print("NOT DIA")
```

STRONG NUMBER

```python
A strong number is a number whose sum of the factorials of its digits equals the original number.
For example, 145 is a strong number because 1!+4!+5!=1+24+120=145.
```

```python
def factoriall(n):
    fact = 1
    while(n != 0):
        fact = fact * (n)
        n = n - 1
    return fact
n = int(input("ENTER N : "))
num = 0
a = n
while(n != 0):
    num = num + factoriall(n%10)
    n = n//10
if num == a:
    print("STRONG")
else:
    print("NOT A STRONG") 
```

PRIME NUMBER

```python
1 is not a prime number
negative numbersa are not prime number
2 is smallest prime number
```

```python
from math import sqrt
def prime(n):
    if n < 2:
        return "NOT PRIME"
    i = 2
    while(i <= int(sqrt(n))):
        if n%i == 0:
            return "NOT PRIME"
        i = i + 1
    return "PRIME"
n = int(input("ENTER N : "))
print(prime(n))
```

TO CHECK A NUMBER COMES IN FIBONACCI SERIES OR NOT

```python
def main(n):
    if n == 0 or n == 1:
        return True
    else:
        a = 0
        b = 1
        while(True):
            c = a + b
            if c == n:
                return True
            if c > n:
                return False
            a = b
            b = c
n = int(input("ENTER N : "))
print(main(n))
```
