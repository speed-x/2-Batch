MAX LENGTH OF A SUBARRAY WHOSE SUM == 0

```python
arr = [15, -2, 2, -8, 1, 7, 10, 23]
n = len(arr)-1
presum = {}
sum = 0
max_len = 0
for i in range(0,n):
    sum = sum + arr[i]
    if sum == 0:
        max_len = i + 1
    if sum in presum:
        max_len = max(max_len,i - presum[sum])
    else:
        presum[sum] = i
print(max_len)
```

COUNT NUMBER OF PAIRS WHOSE SUM == K

```python
arr = [1,5,7,-1,5]
k = 6
count = 0
d = {}
for i in arr:
    complement = k - i
    if complement in d:
        count = count + d[complement]
    d[i] = d.get(i,0) + 1
print(count)
```

GIVEN THE ARRAY IS SORTED FIND IF THE THERE EXISTS A PAIR WITH SUM == k

```python
def main(arr,k):
    st = 0
    end = len(arr) - 1
    while st < end:
        sum = arr[st] + arr[end]
        if sum == k:
            return True
        elif sum > k:
            end = end - 1
        else:
            st = st + 1
    return False
n = int(input())
arr = list(map(int,input().split()))
k = int(input())
print(main(arr,k))
        
```

GIVEN THE ARRAY IS SORTED FIND IF THE THERE EXISTS A PAIR WITH PRODUCT == k

```python
def main(arr,k):
    st = 0
    end = len(arr) - 1
    while st < end:
        sum = arr[st] * arr[end]
        if sum == k:
            return (arr[st],arr[end])
        elif sum > k:
            end = end - 1
        else:
            st = st + 1
    return False
n = int(input())
arr = list(map(int,input().split()))
k = int(input())
a = main(arr,k)
print(a)
```

```python
def rotateby_k(arr,st,end):
    while(st < end):
        arr[st],arr[end]=arr[end],arr[st]
        st = st + 1
        end = end - 1
    return arr
n = int(input())
arr = list(map(int,input().split()))
k = int(input())
if k > n:
    k = k % n
rotateby_k(arr,0,n-1)
rotateby_k(arr,0,k-1)
rotateby_k(arr,k,n-1)
print(arr)
```

FACTORIAL USING RECURSSION

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
n = int(input())
print(factorial(n))
```

FIBONACCI SERIES USING RECURSSION

```python
def n_th_fibo(n):
    if n <= 1:
        return n
    return n_th_fibo(n-1) + n_th_fibo(n-2)
n = 10
print(n_th_fibo(n))
```

GCD USING RECURSION OR HCF

```python
def hcf(a,b):
    if b == 0:
        return a
    return hcf(b,a%b)
a = int(input())
b = int(input())
x = hcf(a,b)
print(x)
```
