```python
# -------------------- BASIC LOGIC --------------------

# 1. Swap two variables using a third variable
def swap_with_third(a, b):
    temp = a
    a = b
    b = temp
    return a, b

# 2. Swap two variables without a third variable
def swap_without_third(a, b):
    a, b = b, a
    return a, b

# 3. Swap two variables using Python (pythonic way)
def swap_pythonic(a, b):
    return b, a

# 4. Swap three variables
def swap_three(a, b, c):
    a, b, c = c, a, b
    return a, b, c

# 5. Check even or odd
def is_even(n):
    return n % 2 == 0

# 6. Check positive or negative
def is_positive(n):
    return n > 0

# 7. Check leap year
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# 8. Check integer or decimal
def is_integer(num):
    return num == int(num)

# 9. Find square root
def sqrt(n):
    return n ** 0.5

# 10. Find power of a number
def power(x, y):
    return x ** y

# 11. Print multiplication table up to n
def multiplication_table(n):
    return [n * i for i in range(1, 11)]

# 12. Print natural numbers from 1 to n
def natural_numbers(n):
    return list(range(1, n + 1))

# 13. Print even numbers from 1 to n
def even_numbers(n):
    return [i for i in range(2, n + 1, 2)]

# 14. Print odd numbers from 1 to n
def odd_numbers(n):
    return [i for i in range(1, n + 1, 2)]

# 15. Sum of numbers from 1 to n
def sum_n(n):
    return n * (n + 1) // 2

# 16. Sum of squares from 1 to n
def sum_of_squares(n):
    return n * (n + 1) * (2 * n + 1) // 6

# 17. Sum of cubes from 1 to n
def sum_of_cubes(n):
    total = n * (n + 1) // 2
    return total * total

# 18. Sum of all factors of n
def sum_of_factors(n):
    return sum(i for i in range(1, n + 1) if n % i == 0)

# 19. Count digits in a number
def count_digits(n):
    return len(str(abs(n)))

# 20. Sum of digits in a number
def sum_of_digits(n):
    return sum(int(d) for d in str(abs(n)))

# 21. Reverse a number
def reverse_number(n):
    return int(str(abs(n))[::-1]) * (-1 if n < 0 else 1)

# 22. Check if number is a palindrome
def is_palindrome(n):
    s = str(n)
    return s == s[::-1]

# 23. Sum of number and its reverse
def sum_with_reverse(n):
    return n + reverse_number(n)

# 24. Check if number is an Armstrong number
def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    return sum(int(d) ** power for d in digits) == n

# 25. Diagonal sum of a square matrix
def diagonal_sum(matrix):
    return sum(matrix[i][i] + matrix[i][-i-1] for i in range(len(matrix))) if matrix else 0

# 26. Factorial of a number
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# 27. Product of digits in a number
def product_of_digits(n):
    product = 1
    for d in str(abs(n)):
        product *= int(d)
    return product

# 28. Check if a number is a spy number
def is_spy_number(n):
    digits = [int(d) for d in str(abs(n))]
    return sum(digits) == product_of_digits(n)

# 29. Check if a number is a perfect number
def is_perfect_number(n):
    return sum(i for i in range(1, n) if n % i == 0) == n

# 30. Check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# 31. Print first n terms of Fibonacci series
def fibonacci_series(n):
    if n <= 0:
        return []
    a, b = 0, 1
    series = [a]
    for _ in range(1, n):
        a, b = b, a + b
        series.append(a)
    return series

# 32. Check if a number is in Fibonacci series
def is_in_fibonacci(num):
    a, b = 0, 1
    while b <= num:
        if b == num:
            return True
        a, b = b, a + b
    return num == 0

# 33. Check if a number is a Sunny number
def is_sunny_number(n):
    return ((n + 1) ** 0.5).is_integer()

# 34. Check if a number is a Happy number
def is_happy_number(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(d)**2 for d in str(n))
    return n == 1


# -------------------- ARRAYS & STRINGS --------------------

# 1. Take input line-wise and space-separated
def input_line_space_separated(s):
    return list(map(int, s.strip().split()))

# 2. Sum of array
def sum_array(arr):
    return sum(arr)

# 3. Max element in array
def max_in_array(arr):
    return max(arr)

# 4. Min element in array
def min_in_array(arr):
    return min(arr)

# 5. Average of array
def average_array(arr):
    return sum(arr) / len(arr) if arr else 0

# 6. Sum of all prime numbers in array
def sum_of_primes_in_array(arr):
    return sum(x for x in arr if is_prime(x))

# 7. Count prime numbers in array
def count_primes_in_array(arr):
    return sum(1 for x in arr if is_prime(x))

# 8. Linear search
def linear_search(arr, target):
    return target in arr

# 9. Binary search (sorted array)
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return True
        elif arr[mid] < target:
            left += 1
        else:
            right -= 1
    return False

# 10. Reverse a string
def reverse_string(s):
    return s[::-1]

# 11. Reverse an array
def reverse_array(arr):
    return arr[::-1]

# 12. Selection sort
def selection_sort(arr):
    a = arr[:]
    n = len(a)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
    return a

# 13. Bubble sort
def bubble_sort(arr):
    a = arr[:]
    n = len(a)
    for i in range(n):
        for j in range(0, n-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a

# 14. Sum of pairs == k
def pair_sum_equals_k(arr, k):
    seen = set()
    for num in arr:
        if k - num in seen:
            return True
        seen.add(num)
    return False

# 15. Second highest element
def second_highest(arr):
    unique = list(set(arr))
    if len(unique) < 2:
        return None
    unique.sort(reverse=True)
    return unique[1]

# 16. Second lowest element
def second_lowest(arr):
    unique = list(set(arr))
    if len(unique) < 2:
        return None
    unique.sort()
    return unique[1]

# 17. Anagram strings
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)

# 18. Reverse word order: "I love you" → "you love I"
def reverse_word_order(s):
    return ' '.join(s.strip().split()[::-1])

# 19. Reverse each word: "I love you" → "I evol uoy"
def reverse_each_word(s):
    return ' '.join(word[::-1] for word in s.strip().split())

# 20. Count special characters
def count_special_characters(s):
    return sum(1 for ch in s if not ch.isalnum() and not ch.isspace())

# 21. Sort string with characters and digits separately
def sort_mixed_string(s):
    chars = sorted([c for c in s if c.isalpha()])
    digits = sorted([c for c in s if c.isdigit()])
    return ''.join(chars + digits)

# 22. ASCII value sum
def ascii_sum(s):
    return sum(ord(c) for c in s)

# 23. "abcd" → 1+2+3+4 = 10
def alphabet_position_sum(s):
    return sum(ord(c) - ord('a') + 1 for c in s.lower() if c.isalpha())

# 24. "abcd" → "zyxw"
def reverse_alphabet_map(s):
    return ''.join(chr(ord('z') - (ord(c) - ord('a'))) for c in s.lower() if c.isalpha())

# 25. Replace vowel with next vowel and consonant with next consonant
def replace_next_vowel_consonant(s):
    vowels = 'aeiou'
    next_vowel = {v: vowels[(i + 1) % len(vowels)] for i, v in enumerate(vowels)}
    def next_char(c):
        if c.lower() in vowels:
            return next_vowel[c.lower()]
        for i in range(1, 26):
            nc = chr((ord(c.lower()) - ord('a') + i) % 26 + ord('a'))
            if nc not in vowels:
                return nc
    return ''.join(next_char(c) if c.isalpha() else c for c in s.lower())

# 26. "g3v2k5" → "gggvvkkkkk"
def expand_compressed_string(s):
    result = ''
    i = 0
    while i < len(s):
        if s[i].isalpha() and i + 1 < len(s) and s[i+1].isdigit():
            result += s[i] * int(s[i+1])
            i += 2
        else:
            result += s[i]
            i += 1
    return result

# 27. Largest number in string
def largest_number_in_string(s):
    import re
    numbers = re.findall(r'\d+', s)
    return max(map(int, numbers)) if numbers else None

# 28. Count vowels in array of strings
def count_vowels_in_array(arr):
    vowels = 'aeiouAEIOU'
    return sum(1 for word in arr for ch in word if ch in vowels)

# 29. Sum of even and odd numbers and absolute difference
def even_odd_sum_difference(arr):
    even = sum(x for x in arr if x % 2 == 0)
    odd = sum(x for x in arr if x % 2 != 0)
    return abs(even - odd)

# 30. Remove duplicates from sorted array
def remove_duplicates_sorted(arr):
    return list(dict.fromkeys(arr))

# 31. Print all substrings
def all_substrings(s):
    return [s[i:j] for i in range(len(s)) for j in range(i + 1, len(s) + 1)]


# -------------------- HASHMAPS --------------------

# 1. Remove duplicates from array
def remove_duplicates(arr):
    return list(dict.fromkeys(arr))

# 2. Remove duplicates from string
def remove_duplicates_from_string(s):
    seen = set()
    return ''.join([ch for ch in s if not (ch in seen or seen.add(ch))])

# 3. Frequency of each character in string
def char_frequency(s):
    from collections import Counter
    return dict(Counter(s))

# 4. Remove duplicates (general)
def remove_duplicates_general(arr):
    return list(dict.fromkeys(arr))

# 5. Find duplicates
def find_duplicates(arr):
    from collections import Counter
    return [item for item, count in Counter(arr).items() if count > 1]

# 6. Frequency descending order
def frequency_descending(arr):
    from collections import Counter
    return sorted(Counter(arr).items(), key=lambda x: (-x[1], x[0]))

# 7. Frequency ascending order
def frequency_ascending(arr):
    from collections import Counter
    return sorted(Counter(arr).items(), key=lambda x: (x[1], x[0]))


# -------------------- MISCELLANEOUS --------------------

# 1. Max length subarray with sum = 0
def max_len_subarray_sum_zero(arr):
    prefix_sum = 0
    seen = {0: -1}
    max_len = 0
    for i, num in enumerate(arr):
        prefix_sum += num
        if prefix_sum in seen:
            max_len = max(max_len, i - seen[prefix_sum])
        else:
            seen[prefix_sum] = i
    return max_len

# 2. Count pairs with sum == k
def count_pairs_with_sum_k(arr, k):
    from collections import Counter
    count = 0
    freq = Counter(arr)
    for num in arr:
        count += freq[k - num]
        if k - num == num:
            count -= 1
    return count // 2

# 3. Pair with sum == k in sorted array
def has_pair_with_sum_sorted(arr, k):
    left, right = 0, len(arr) - 1
    while left < right:
        s = arr[left] + arr[right]
        if s == k:
            return True
        elif s < k:
            left += 1
        else:
            right -= 1
    return False

# 4. Pair with product == k in sorted array
def has_pair_with_product_sorted(arr, k):
    left, right = 0, len(arr) - 1
    while left < right:
        p = arr[left] * arr[right]
        if p == k:
            return True
        elif p < k:
            left += 1
        else:
            right -= 1
    return False

# 5. Rotate array by k
def rotate_array(arr, k):
    k %= len(arr)
    return arr[-k:] + arr[:-k]

# 6. Factorial using recursion
def factorial_recursive(n):
    return 1 if n <= 1 else n * factorial_recursive(n - 1)

# 7. Fibonacci series using recursion
def fibonacci_recursive(n):
    def fib(i):
        return i if i <= 1 else fib(i - 1) + fib(i - 2)
    return [fib(i) for i in range(n)]

# 8. GCD using recursion
def gcd_recursive(a, b):
    return a if b == 0 else gcd_recursive(b, a % b)

# 9. Tower of Hanoi with 4 variables
def tower_of_hanoi(n, source, aux1, aux2, target, steps=None):
    if steps is None:
        steps = []
    if n == 0:
        return steps
    if n == 1:
        steps.append((source, target))
    else:
        tower_of_hanoi(n - 2, source, target, aux2, aux1, steps)
        steps.append((source, aux2))
        steps.append((source, target))
        steps.append((aux2, target))
        tower_of_hanoi(n - 2, aux1, source, aux2, target, steps)
    return steps

# 10. Is number in Fibonacci using recursion
def is_fibonacci_recursive(n, a=0, b=1):
    if n == a or n == b:
        return True
    if b > n:
        return False
    return is_fibonacci_recursive(n, b, a + b)
```
