```python
from math import sqrt
def prime(n):
    if n < 2:
        return False
    i = 2
    end = int(sqrt(n))
    while(i <= end):
        if n%i == 0 :
            return False
    return True
def arms(n):
    sum = 0
    count = 0
    a = n
    while(n):
        n = n//10
        count = count + 1
    n = a
    while(n):
        b = n%10
        sum = sum + (b**count)
        n = n//10
    if sum == a:
        return True
    return False
def reverse(n):
    rev = 0
    while(n):
        b = n%10
        rev = (rev*10)+b
        n = n//10
    return rev
n = int(input("ENTER N : "))
arr = list(map(int,input().split()))
sum_even = 0
sum_odd = 0
for i in arr:
    if i%2 == 0:
        sum_even = sum_even + i
    else:
        sum_odd = sum_odd + i
difference = abs(sum_odd-sum_even)
if difference%2 != 0:
    if prime(difference) == True:
        print("True")
    else:
        if arms(difference) == True:
            print("True")
        else:
            if difference%2 == 0:
                print("True")
            else:
                print("False")
else:
    r = reverse(difference)
    r = r + 5
    if prime(r) == True:
        print("True")
    else:
        print("False")
```
