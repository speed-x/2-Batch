```python
FIND AVERAGE OF AN ARRAY 
if Aver is int check it is arms
else if float convert it into int and check if it is prime
```

```python
from math import sqrt
def average(n,arr):
    sum = 0
    for i in arr:
        sum = sum + i
    a = (sum)/n
    return a
def arms(c):
    count = 0
    a = c
    while(c):
        c = c//10
        count = count + 1
    c = a
    sum = 0
    while(c):
        b = c%10
        sum = sum + (b**count)
        c = c//10
    if sum == a:
        return True
    return False
def prime(c):
    if c < 2:
        return False
    end = int(sqrt(c))
    i = 2
    while(i <= end):
        if c%i == 0:
            return False
    return True    

n = int(input("ENTER N : "))
arr = list(map(int,input().split()))
c = average(n,arr)
if type(c) == int:
    print(arms(c))
else:
    c = int(c)
    print(prime(c))
```
