SUM OF ALL PRIME NUMBERS IN AN ARRAY

```python
from math import sqrt
def prime(n):
    if n < 2:
        return False
    i = 2
    end = int(sqrt(n))
    while(i <= end):
        if n%i == 0:
            return False
        i = i + 1
    return True
n = int(input("ENTER N : "))
arr = list(map(int,input().split()))
sum = 0
for i in arr:
    x = prime(i)
    if x == True:
        sum = sum + i
print(sum)
```
